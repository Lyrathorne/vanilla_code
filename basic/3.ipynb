```python
a = []
s = 1
while True:
          b = input("Введи че нибудь")
          
          if b.isdigit() == True:
                    a.append(int(b))
          elif b == "+":
                    print(sum(a))
          elif b == "*":
                    for i in a:
                              s = s * i
                    print(s)
          elif b == "end":
                    print(*a)
                    break
          else:
                    print("Тупое уебище")
                    break
```
