Создай несколько объектов Path и для каждого выведи name, stem, suffix, parent. Примеры: data/users.json, /var/log/app.log, archive.tar.gz.


```python
from pathlib import Path
paths = [Path("Итоговая версия Заявки.py"), Path("Untitled-2.ipynb"), Path("3.ipynb")]
for p in paths:          
          print(p)
          print(p.name)
          print(p.stem)
          print(p.suffix)
          print(p.parent)
```

Создай словарь пользователя, преврати его в JSON-строку, сохрани в файл, прочитай файл обратно и выведи имя и число ролей.


```python
import json
from pathlib import Path
user = {
          "name": "Alex",
          "age": 25,
          "roles": ["admin", "user"]
}
text = json.dumps(user)
file = Path("file.json")
file.write_text(text)
data = json.loads(json.dumps(user))
print(data["name"])
print(data["roles"])


```

Получай текущее UTC-время, прибавляй 90 минут, выводи оба значения и сериализуй второе через isoformat().


```python
import datetime
time_now = datetime.datetime.now(datetime.timezone.utc)
time = time_now + datetime.timedelta(days = 0, hours = 1, minutes = 30)
print(time_now)
print(time.isoformat())
```
