}

Ответ:

{"status": "sent"}

```python
from fastapi import FastAPI
from pydantic import BaseModel
app = FastAPI()
class MessageRequest(BaseModel):
          username: str
          text: str
          
@app.post("/send")
def send_message(data: MessageRequest):
        return{
          "user": data.username,
          "text": data.text,
          "status": "sent"
        }
```

Задание 6. Хранение сообщений

Создай список:

messages = []

При каждом POST добавляй туда сообщение

```python
from fastapi import FastAPI
from pydantic import BaseModel
app = FastAPI()
messages = []
class MessageRequest(BaseModel):
          username: str
          text: str
          
@app.post("/send")
def send_message(data: MessageRequest):
        messages.append(data.dict())
        return{
          
          "status": "sent"
        }
        
```
