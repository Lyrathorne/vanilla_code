Найдите сумму всех чисел меньше 1000, кратных 3 или 5.

```python
s = [i for i in range(1,1000) if i % 3 == 0 or i % 5 == 0]
print(sum(s))
```

Напишите программу которая считает сколько раз каждая буква встречается в слове. Используйте словарь для хранения количества.

```python
word = input("Введите слово: ")
letters = {}
for letter in word:   
    if letter in letters:
        letters[letter] += 1
    else:
        letters[letter] = 1
print(letters)
```

Добавьте меню (while True) чтобы пользователь мог выбирать действия: добавить контакт, найти контакт, удалить контакт, показать все контакты.

```python
contacts = {}
while True:
    print("________Меню________")
    print("1. Добавить контакт")
    print("2. Удалить контакт")
    print("3. Найти контакт")
    print("4. Показать контакты")
    a = int(input("Выберите действие: "))
    
    if a == 1:
        name = input("Введите имя: ")
        number = input("Введите номер: ")
        contacts[name] = number
    elif a == 4:
        print(contacts.items())
    elif a == 2:
        name = input("Введите контакт который хотите удалить: ")
        if name in contacts:
            del contacts[name]
        else:
            print("Контакт не найден")
    elif a == 3:
        name = input("Введите контакт который хотите найти: ")
        if name in contacts:
            print(contacts[name])
        else: 
            print("Контакт не найден")


```
