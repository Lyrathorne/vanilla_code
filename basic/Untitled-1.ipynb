Задание 1 — Пользователь

Создай класс:

User

Требования:

username

age

метод introduce() → "Hi, I'm Alex, 20 years old"

```python
class User:
          def __init__(self, username, age):
                  self.username = username
                  self.age = age
          def inroduce(self):
                  print(f"Hi, I am {self.username}, {self.age} years old")
```

Задание 2 — Сообщение
Message

text

sender

метод preview() → первые 10 символов

```python
class Message:
          def __init__(self, text, sender):
                  self.text = text
                  self.sender = sender
          def preview(self):
                  print(f"{text[:10]} - {sender}")
```
