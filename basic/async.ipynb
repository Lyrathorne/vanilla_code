```python
import time

def task(name):
    print(f"{name} started")
    time.sleep(5)
    print(f"{name} finished")

task("A")
task("B")
task("C")
```

async


```python
import time
import asyncio
async def task(name):
    print(f"{name} started")
    await asyncio.sleep(5)
    print(f"{name} finished") 
async def main():
    asyncio.gather(
    task("A"),
    task("B"),
    task("C"),
    )

await main()
```
